```python
q1 = np.percentile(df["Age"].dropna(), 25)   # 25%
q3 = np.percentile(df["Age"].dropna(), 75)   # 75%
q1,q3
```

```python
IQR = q3 - q1

lower = q1 - 1.5 * IQR
upper = q3 + 1.5 * IQR

lower, upper
```
